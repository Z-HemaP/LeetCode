# 2. Add Two Numbers

You are given two non-empty linked lists representing two non-negative integers. The digits are stored in reverse order, and each of their nodes contains a single digit. Add the two numbers and return the sum as a linked list.

You may assume the two numbers do not contain any leading zero, except the number 0 itself.

```python
from typing import Optional

class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next
```

```python
def addTwoNumbers(l1: Optional[ListNode], l2: Optional[ListNode]) -> Optional[ListNode]:
    dummy = ListNode()
    current = dummy
    carry = 0

    while l1 or l2 or carry:
        val1 = l1.val if l1 else 0
        val2 = l2.val if l2 else 0
        total = val1 + val2 + carry
        carry, digit = divmod(total, 10)

        current.next = ListNode(digit)
        current = current.next

        l1 = l1.next if l1 else None
        l2 = l2.next if l2 else None

    return dummy.next
```

```python
def array_to_linkedlist(arr):
    """Convert a Python list/array to a linked list of ListNode objects."""
    if not arr:
        return None
    head = ListNode(arr[0])
    current = head
    for val in arr[1:]:
        current.next = ListNode(val)
        current = current.next
    return head
```

```python
def build_list(digits):
    dummy = ListNode()
    current = dummy
    for d in digits:
        current.next = ListNode(d)
        current = current.next
    return dummy.next

def list_to_string(node):
    digits = []
    while node:
        digits.append(str(node.val))
        node = node.next
    return " -> ".join(digits)
```

```python
# Demo: Using array_to_linkedlist
arr = [3, 4, 2]
linked_list = array_to_linkedlist(arr)
print(f"Array {arr} -> LinkedList: {list_to_string(linked_list)}")
```

```python
# Example 1: 342 + 465 = 807
l1 = build_list([2, 4, 3])
l2 = build_list([5, 6, 4])
result = addTwoNumbers(l1, l2)
print(list_to_string(result))
```

```python
# Example 2: 0 + 0 = 0
l1 = build_list([0])
l2 = build_list([0])
result = addTwoNumbers(l1, l2)
print(list_to_string(result))
```

```python
# Example 3: 9999999 + 9999 = 10009998
l1 = build_list([9, 9, 9, 9, 9, 9, 9])
l2 = build_list([9, 9, 9, 9])
result = addTwoNumbers(l1, l2)
print(list_to_string(result))
```
