# 2. Add Two Numbers

You are given two non-empty linked lists representing two non-negative integers. The digits are stored in reverse order, and each of their nodes contains a single digit. Add the two numbers and return the sum as a linked list.

You may assume the two numbers do not contain any leading zero, except the number 0 itself.

```python
from typing import Optional

class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next
```

```python
def array_to_linkedlist(arr):
    if not arr:
        return None
    head = ListNode(arr[0])
    current = head
    for val in arr[1:]:
        current.next = ListNode(val)
        current = current.next
    return head
```

```python
class Solution:
    def printList(self,node):
        while node is not None:
            print(f"{node.val}", end="")
            if node.next is not None:
                print("->", end="")
            node = node.next

    def addTwoNumbers(self, l1: Optional[ListNode], l2: Optional[ListNode]) -> Optional[ListNode]:
        head = l1.val + l2.val 
        carry = int(head/10)
        out_list = ListNode(head%10)
        l1 = l1.next
        l2 = l2.next
        current = out_list
        while l1 is not None or l2 is not None or carry:
            if l1 is None:
                l1_val = 0
            else:
                l1_val = l1.val
                l1 = l1.next
            if l2 is None:
                l2_val = 0
            else:
                l2_val = l2.val
                l2 = l2.next
            addn = l1_val + l2_val + carry
            carry = 0
            if addn> 9:
                carry = int(addn/10)
            current.next = ListNode(addn%10)
            current = current.next     
        return out_list
```

```python
# Demo: Using array_to_linkedlist
l1 = array_to_linkedlist([2,4,3])
l2 = array_to_linkedlist([5,6,4])
sol = Solution()
result = sol.addTwoNumbers(l1, l2)
print(sol.printList(result))
```
